```python
%%writefile maximo_lib/asset_utils.py

def normalize_asset_number(asset_number):

    if asset_number is None:
        return ""

    return asset_number.strip().upper()


def is_valid_asset_status(status):


    allowed_statuses = [
        "ACTIVE",
        "INACTIVE",
        "DECOMMISSIONED"
    ]

    return status in allowed_statuses


def create_asset_message(asset_number, status):

    clean_asset_number = normalize_asset_number(asset_number)

    return "Asset " + clean_asset_number + " has status " + status
```

```python
from maximo_lib.asset_utils import normalize_asset_number
from maximo_lib.asset_utils import is_valid_asset_status
from maximo_lib.asset_utils import create_asset_message
```

```python
asset_number = "  pump-100  "
status = "ACTIVE"

print(normalize_asset_number(asset_number))
print(is_valid_asset_status(status))
print(create_asset_message(asset_number, status))
```

```python
%%writefile maximo_lib/validation.py

from maximo_lib.asset_utils import normalize_asset_number
from maximo_lib.asset_utils import is_valid_asset_status


def validate_asset(asset_number, status):

    clean_asset_number = normalize_asset_number(asset_number)

    if clean_asset_number == "":
        return False, "Asset number is required"

    if not is_valid_asset_status(status):
        return False, "Invalid asset status"

    return True, "Asset is valid"
```

```python
from maximo_lib.validation import validate_asset

valid, message = validate_asset(" pump-100 ", "ACTIVE")

print(valid)
print(message)
```

```python
%%writefile automation_scripts/asset_status_script.py

from maximo_lib.asset_utils import normalize_asset_number
from maximo_lib_asset_utils import validate_asset

def process_asset(asset_number,status):
  valid, message = validate_asset(asset_number,status)

  if not valid :
    return message
  clean_asset_number = normalize_asset_number(asset_number)
  return "Asset " + clean_asset_number + "processed sucessfully"
```

```python
from getpass import getpass
import subprocess
import base64

username = input("GitHub username:").strip()
token = getpass("GitHub token: ").strip()

credentials = base64.b64encode(
    f"{username}:{token}".encode()
).decode()

result = subprocess.run(
    [
        "git",
        "-c",
        f"http.extraHeader=Authorization: Basic {credentials}",
        "push",
        "-u",
        "origin",
        "HEAD"
    ],
    text=True,
    capture_output=True
)

print(result.stdout)
print(result.stderr)

del token
del credentials
```
